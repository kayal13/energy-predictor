import libraries

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.tree import DecisionTreeClassifier
from sklearn. linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix
import joblib
```

load dataset

```python
data=pd.read_csv("/content/waste_classification_dataset_500.csv")
print(data.head())
```

feature and target selection

```python
X=data.drop("WasteType",axis=1)
Y=data["WasteType"]
```

encode target labels

```python
encoder=LabelEncoder()
Y=encoder.fit_transform(Y)
Y
```

split dataset

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42)
```

logistic regression model

```python
lr=LogisticRegression(max_iter=500)
lr.fit(X_train,Y_train)
lr_pred=lr.predict(X_test)
lr_pred
```

print accuracy


```python
print("Logistic Regression Accuracy")
print(accuracy_score(Y_test,lr_pred))
print("\nClassification Report")
print(classification_report(Y_test,lr_pred))
print("\nConfusion Matrix")
print(confusion_matrix(Y_test,lr_pred))
```

joblib

```python
import joblib
joblib.dump(lr,'waste_classification.pkl')
```

decision tree model

```python
dt=DecisionTreeClassifier(random_state=42)
dt.fit(X_train,Y_train)
dt_pred=dt.predict(X_test)
```

```python
print("Decision Tree Accuracy")
print(accuracy_score(Y_test,dt_pred))
print("\nClassification Report")
print(classification_report(Y_test,dt_pred))
print("\nConfusion Matrix")
print(confusion_matrix(Y_test,dt_pred))
```

```python
import joblib
joblib.dump(dt,'waste_classification(1).pkl')
```

compare models

```python
dt_accuracy = accuracy_score(Y_test, dt_pred)
lr_accuracy = accuracy_score(Y_test, lr_pred)

print("Decision Tree :", dt_accuracy)

print("Logistic Regression :", lr_accuracy)

if dt_accuracy > lr_accuracy:
    print("Decision Tree performed better.")
elif lr_accuracy > dt_accuracy:
    print("Logistic Regression performed better.")
else:
    print("Both models performed equally.")
```
